# Task 1
Learning the basics of Keras API for TensorFlow

Understanding and Experimenting with MLP

```python
import tensorflow as tf
from tensorflow.keras import layers, models
import numpy as np
```

```python
# Load Fashion MNIST dataset
fashion_mnist = tf.keras.datasets.fashion_mnist
(x_train, y_train), (x_test, y_test) = fashion_mnist.load_data()

# Normalize pixel values
x_train, x_test = x_train / 255.0, x_test / 255.0

# Flatten input data for MLP
x_train = x_train.reshape(-1, 28 * 28)
x_test = x_test.reshape(-1, 28 * 28)

# Define the MLP model
mlp_model = models.Sequential([
    layers.InputLayer(input_shape=(28 * 28,)),
    layers.Dense(128, activation='relu'),
    layers.Dense(64, activation='relu'),
    layers.Dense(10, activation='softmax')
])

# Compile the model
mlp_model.compile(optimizer='adam',
                  loss='sparse_categorical_crossentropy',
                  metrics=['accuracy'])

# Train the model
mlp_model.fit(x_train, y_train, validation_split=0.1, epochs=10)

# Evaluate the model
mlp_model.evaluate(x_test, y_test)
```

Understanding and experimenting with CNN

```python
# Reshape input data for CNN
x_train_cnn = x_train.reshape(-1, 28, 28, 1)
x_test_cnn = x_test.reshape(-1, 28, 28, 1)

# Define the CNN model
cnn_model = models.Sequential([
    layers.Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1)),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.Flatten(),
    layers.Dense(64, activation='relu'),
    layers.Dense(10, activation='softmax')
])

# Compile the model
cnn_model.compile(optimizer='adam',
                  loss='sparse_categorical_crossentropy',
                  metrics=['accuracy'])

# Train the model
cnn_model.fit(x_train_cnn, y_train, validation_split=0.1, epochs=10)

# Evaluate the model
cnn_model.evaluate(x_test_cnn, y_test)
```

# Task 2
Develop a “Tell-the-time” network

Data Preparation

```python
import numpy as np
from sklearn.model_selection import train_test_split
from tensorflow.keras.utils import to_categorical
import tensorflow as tf
from tensorflow.keras.models import Model, Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout, BatchNormalization, Input, GlobalAveragePooling2D
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau, LearningRateScheduler
from tensorflow.keras.preprocessing.image import ImageDataGenerator
import matplotlib.pyplot as plt
import os
```

```python


# Load the data
images = np.load('/home/s4022297/idl/images.npy')
labels = np.load('/home/s4022297/idl/labels.npy')

# Normalize the images
images = images.astype('float32') / 255.0

# Split the data
X_train, X_test, y_train, y_test = train_test_split(images, labels, test_size=0.2, random_state=42)

# Further split training data into training and validation
X_train, X_val, y_train, y_val = train_test_split(X_train, y_train, test_size=0.125, random_state=42)  # 0.125 x 0.8 = 0.1

# Function to convert time to minutes past midnight
def time_to_minutes(time):
    return time[:, 0] * 60 + time[:, 1]

# Prepare labels for classification approach (10-minute intervals)
y_train_minutes = time_to_minutes(y_train)
y_val_minutes = time_to_minutes(y_val)
y_test_minutes = time_to_minutes(y_test)

# Use 10-minute intervals to classify the labels
y_train_class = y_train_minutes // 10
y_val_class = y_val_minutes // 10
y_test_class = y_test_minutes // 10

# Set the number of classes to match 10-minute intervals in 12 hours (72 classes)
num_classes = 72  # 72 classes representing 10-minute intervals

# Convert labels to categorical
y_train_cat = to_categorical(y_train_class, num_classes=num_classes)
y_val_cat = to_categorical(y_val_class, num_classes=num_classes)
y_test_cat = to_categorical(y_test_class, num_classes=num_classes)

# Add channel dimension to the image data to make it 4D
X_train = X_train.reshape(X_train.shape[0], X_train.shape[1], X_train.shape[2], 1)
X_val = X_val.reshape(X_val.shape[0], X_val.shape[1], X_val.shape[2], 1)
X_test = X_test.reshape(X_test.shape[0], X_test.shape[1], X_test.shape[2], 1)

# For regression approach
y_train_reg = y_train_minutes / 720  # Normalize to [0, 1]
y_val_reg = y_val_minutes / 720
y_test_reg = y_test_minutes / 720

# Verify the new shapes
print("Data preparation completed.")
print(f"Training set shape: {X_train.shape}")
print(f"Validation set shape: {X_val.shape}")
print(f"Test set shape: {X_test.shape}")
print(f"Training labels shape: {y_train_cat.shape}")
print(f"Validation labels shape: {y_val_cat.shape}")
print(f"Test labels shape: {y_test_cat.shape}")
```

Preparing the Model Architecture

```python

# Set environment variable to use GPU 0 
os.environ["CUDA_VISIBLE_DEVICES"] = "1"

# Set GPU memory growth to prevent full allocation at startup
gpus = tf.config.list_physical_devices('GPU')
if gpus:
    try:
        for gpu in gpus:
            tf.config.experimental.set_memory_growth(gpu, True)
    except RuntimeError as e:
        print(e)

# Set up MirroredStrategy to utilize GPU 0 only
strategy = tf.distribute.MirroredStrategy(devices=["/gpu:1"])

def create_base_model(input_shape, output_units, activation='softmax'):
    inputs = Input(shape=input_shape)
    
    x = Conv2D(32, (3, 3), activation='relu', padding='same')(inputs)
    x = MaxPooling2D((2, 2))(x)
    x = Conv2D(64, (3, 3), activation='relu', padding='same')(x)
    x = MaxPooling2D((2, 2))(x)
    x = Conv2D(64, (3, 3), activation='relu', padding='same')(x)
    x = MaxPooling2D((2, 2))(x)
    
    x = Flatten()(x)
    x = Dense(64, activation='relu')(x)
    x = Dropout(0.5)(x)
    
    outputs = Dense(output_units, activation=activation)(x)
    
    model = Model(inputs=inputs, outputs=outputs)
    return model
```

Classification Approach

```python
# Assuming data preparation is done and X_train, y_train, etc. are available
# Use 10-minute intervals for more granularity
y_train_class = y_train_minutes // 10
y_val_class = y_val_minutes // 10
y_test_class = y_test_minutes // 10

# Set the number of classes to match 10-minute intervals in 12 hours (72 classes)
num_classes = 72  # 72 classes representing 10-minute intervals

# Convert labels to categorical
y_train_cat = to_categorical(y_train_class, num_classes=num_classes)
y_val_cat = to_categorical(y_val_class, num_classes=num_classes)
y_test_cat = to_categorical(y_test_class, num_classes=num_classes)

# Compute class weights to balance classes
from sklearn.utils import class_weight
class_weights = class_weight.compute_class_weight('balanced', classes=np.unique(y_train_class), y=y_train_class)
class_weights_dict = dict(enumerate(class_weights))

# Slightly more complex model for better learning capacity
model = Sequential([
    Conv2D(32, (3, 3), activation='relu', padding='same', input_shape=(75, 75, 1)),
    BatchNormalization(),
    MaxPooling2D((2, 2)),
    Conv2D(64, (3, 3), activation='relu', padding='same'),
    BatchNormalization(),
    MaxPooling2D((2, 2)),
    Conv2D(128, (3, 3), activation='relu', padding='same'),
    BatchNormalization(),
    MaxPooling2D((2, 2)),
    Conv2D(256, (3, 3), activation='relu', padding='same'),
    BatchNormalization(),
    MaxPooling2D((2, 2)),
    Flatten(),
    Dense(512, activation='relu'),
    BatchNormalization(),
    Dropout(0.5),
    Dense(256, activation='relu'),
    BatchNormalization(),
    Dropout(0.5),
    Dense(num_classes, activation='softmax')
])



# Compile the model
model.compile(optimizer=Adam(learning_rate=0.001),  # Higher initial learning rate
              loss='categorical_crossentropy',
              metrics=['accuracy'])



datagen = ImageDataGenerator(
    rotation_range=5,  # Reduced from 10
    width_shift_range=0.05,
    height_shift_range=0.05,
    zoom_range=0.05,
    horizontal_flip=True
)



# Callbacks
early_stopping = EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True)
reduce_lr = ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=5, min_lr=1e-6)


# Train the model with class weights
history = model.fit(datagen.flow(X_train, y_train_cat, batch_size=32),
                    steps_per_epoch=len(X_train) // 32,
                    epochs=100,
                    validation_data=(X_val, y_val_cat),
                    callbacks=[early_stopping, reduce_lr])

# Function to calculate "common sense" error with correct interval
def common_sense_error(y_true_class, y_pred_class, interval=10):
    y_true_minutes = y_true_class * interval  # Convert class back to minutes
    y_pred_minutes = y_pred_class * interval
    diff = np.abs(y_true_minutes - y_pred_minutes)
    return np.minimum(diff, 720 - diff).mean()

test_loss, test_acc = model.evaluate(X_test, y_test_cat)
print(f"Test accuracy: {test_acc:.4f}")

# Calculate "common sense" error
y_pred = model.predict(X_test)
y_pred_class = np.argmax(y_pred, axis=1)
common_sense_err = common_sense_error(y_test_class, y_pred_class)
print(f"Common sense error: {common_sense_err:.2f} minutes")

```

Regression Approach

```python


# Assuming the input data is correctly normalized
X_train *= 255.0
X_val *= 255.0
X_test *= 255.0

X_train = X_train.reshape(-1, 75, 75, 1)
X_val = X_val.reshape(-1, 75, 75, 1)
X_test = X_test.reshape(-1, 75, 75, 1)

y_train_sin = np.sin(2 * np.pi * y_train_reg)
y_train_cos = np.cos(2 * np.pi * y_train_reg)
y_val_sin = np.sin(2 * np.pi * y_val_reg)
y_val_cos = np.cos(2 * np.pi * y_val_reg)
y_test_sin = np.sin(2 * np.pi * y_test_reg)
y_test_cos = np.cos(2 * np.pi * y_test_reg)

y_train_combined = np.stack([y_train_sin, y_train_cos], axis=-1)
y_val_combined = np.stack([y_val_sin, y_val_cos], axis=-1)
y_test_combined = np.stack([y_test_sin, y_test_cos], axis=-1)

def create_cyclic_model(input_shape):
    model = Sequential([
        Conv2D(32, (3, 3), activation='relu', input_shape=input_shape, padding='same'),
        BatchNormalization(),
        MaxPooling2D((2, 2)),
        Conv2D(64, (3, 3), activation='relu', padding='same'),
        BatchNormalization(),
        MaxPooling2D((2, 2)),
        Conv2D(128, (3, 3), activation='relu', padding='same'),
        BatchNormalization(),
        MaxPooling2D((2, 2)),
        Conv2D(256, (3, 3), activation='relu', padding='same'),
        BatchNormalization(),
        MaxPooling2D((2, 2)),
        Conv2D(512, (3, 3), activation='relu', padding='same'),
        BatchNormalization(),
        GlobalAveragePooling2D(),
        Dense(512, activation='relu', kernel_regularizer=tf.keras.regularizers.l2(0.001)),
        Dropout(0.5),
        Dense(128, activation='relu', kernel_regularizer=tf.keras.regularizers.l2(0.001)),
        Dropout(0.3),
        Dense(2, activation='linear')  # Predicting sin and cos directly
    ])
    return model

def scheduler(epoch, lr):
    if epoch < 50:
        return lr
    else:
        return lr * tf.math.exp(-0.05)

model = create_cyclic_model((75, 75, 1))
model.compile(optimizer=Adam(learning_rate=0.001),
              loss='mse',  # Mean Squared Error
              metrics=['mae'])

datagen = ImageDataGenerator(
    rotation_range=40,
    width_shift_range=0.4,
    height_shift_range=0.4,
    zoom_range=0.4,
    horizontal_flip=True,
    fill_mode='nearest'
)

early_stopping = EarlyStopping(patience=30, restore_best_weights=True)
lr_scheduler = LearningRateScheduler(scheduler)
reduce_lr = ReduceLROnPlateau(monitor='val_loss', factor=0.2, patience=5, min_lr=1e-6)

history = model.fit(datagen.flow(X_train, y_train_combined, batch_size=256),
                    validation_data=(X_val, y_val_combined),
                    epochs=100,
                    callbacks=[early_stopping, lr_scheduler, reduce_lr])

# Prediction and error calculation as previously defined
y_pred = model.predict(X_test)  # No clipping needed unless you observe output beyond [-1, 1]
y_pred_sin = y_pred[:, 0]
y_pred_cos = y_pred[:, 1]

# Conversion from sine and cosine to time
y_pred_angle = np.arctan2(y_pred_sin, y_pred_cos)
y_pred_minutes = (y_pred_angle % (2 * np.pi)) * (720 / (2 * np.pi))

# Define the common sense error calculation function
def common_sense_error_trig(y_true, y_pred):
    error = np.abs(y_true * 720 - y_pred)
    error = np.minimum(error, 720 - error)
    return np.mean(error)

# Calculate and display the common sense error
common_sense_err = common_sense_error_trig(y_test_reg, y_pred_minutes)
print(f"Common sense error: {common_sense_err:.2f} minutes")
```

Multi Head Model

```python
def create_multi_head_model(input_shape):
    inputs = Input(shape=input_shape)
    
    x = Conv2D(32, (3, 3), activation='relu', padding='same')(inputs)
    x = BatchNormalization()(x)
    x = MaxPooling2D((2, 2))(x)
    
    x = Conv2D(64, (3, 3), activation='relu', padding='same')(x)
    x = BatchNormalization()(x)
    x = MaxPooling2D((2, 2))(x)
    
    x = Conv2D(128, (3, 3), activation='relu', padding='same')(x)
    x = BatchNormalization()(x)
    x = MaxPooling2D((2, 2))(x)
    
    x = Conv2D(128, (3, 3), activation='relu', padding='same')(x)
    x = BatchNormalization()(x)
    x = MaxPooling2D((2, 2))(x)
    
    x = Flatten()(x)
    x = Dense(128, activation='relu')(x)
    x = BatchNormalization()(x)
    x = Dropout(0.5)(x)
    
    # Hour head (classification)
    hour_output = Dense(12, activation='softmax', name='hour_output')(x)
    
    # Minute head (regression)
    minute_output = Dense(1, activation='linear', name='minute_output')(x)
    
    model = Model(inputs=inputs, outputs=[hour_output, minute_output])
    return model

# Assuming data preparation is done and input data is in correct shape
if len(X_train.shape) == 3:
    X_train = X_train.reshape(-1, 75, 75, 1)
    X_val = X_val.reshape(-1, 75, 75, 1)
    X_test = X_test.reshape(-1, 75, 75, 1)

# Create and compile the model
model = create_multi_head_model((75, 75, 1))
model.compile(optimizer=Adam(learning_rate=0.001),
              loss={'hour_output': 'sparse_categorical_crossentropy',
                    'minute_output': 'mse'},
              loss_weights={'hour_output': 1.0, 'minute_output': 0.5},
              metrics={'hour_output': 'accuracy', 'minute_output': 'mae'})

# Prepare labels for multi-head model
y_train_hour = y_train[:, 0]
y_train_minute = y_train[:, 1] / 60.0  # Normalize minutes to [0, 1]
y_val_hour = y_val[:, 0]
y_val_minute = y_val[:, 1] / 60.0
y_test_hour = y_test[:, 0]
y_test_minute = y_test[:, 1] / 60.0

# Train the model
early_stopping = EarlyStopping(monitor='val_loss', patience=15, restore_best_weights=True)
reduce_lr = ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=5, min_lr=1e-6)

history = model.fit(X_train, 
                    {'hour_output': y_train_hour, 'minute_output': y_train_minute},
                    validation_data=(X_val, {'hour_output': y_val_hour, 'minute_output': y_val_minute}),
                    epochs=100,
                    batch_size=64,
                    callbacks=[early_stopping, reduce_lr])

# Evaluate the model
test_loss, hour_loss, minute_loss, hour_acc, minute_mae = model.evaluate(
    X_test, {'hour_output': y_test_hour, 'minute_output': y_test_minute}
)
print(f"Test hour accuracy: {hour_acc:.4f}")
print(f"Test minute MAE: {minute_mae:.4f}")

# Function to calculate "common sense" error for multi-head model
def common_sense_error_multi_head(y_true_hour, y_true_minute, y_pred_hour, y_pred_minute):
    y_true_minutes = y_true_hour * 60 + y_true_minute * 60
    y_pred_minutes = np.argmax(y_pred_hour, axis=1) * 60 + y_pred_minute.flatten() * 60
    diff = np.abs(y_true_minutes - y_pred_minutes)
    return np.minimum(diff, 720 - diff).mean()

# Calculate "common sense" error
y_pred_hour, y_pred_minute = model.predict(X_test)
common_sense_err = common_sense_error_multi_head(y_test_hour, y_test_minute, y_pred_hour, y_pred_minute)
print(f"Common sense error: {common_sense_err:.2f} minutes")
```
